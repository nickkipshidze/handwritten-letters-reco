```python
import joblib
import matplotlib.pyplot as plt
from random import randint as rint
```

## Loading the dataset

I made this dataset when I was 15 years old. Not the smartest. So it is in joblib format. It's basically a Python dictionary with two keys: "data" and "labels".

The data values are the normalized images of a handwritten Georgian letter, and the label values are numbers from 0 to 32, 0 for letter ა and 32 for the letter ჰ.

```python
alphabet = [chr(num) for num in range(ord("ა"), ord("ჰ")+1)]
dataset = joblib.load("./dataset.joblib", mmap_mode="r")
```

```python
def plot_data_idx(index, title=False):
    if title:
        plt.title(f"Label: {alphabet[dataset['labels'][index]]}")
    plt.imshow(dataset["data"][index])
    plt.axis(False)

plt.figure(figsize=(7, 4))
plt.title("ნიკას ხელნაწერი ქართული ასოების მონაცემთა ბაზა!", size=12)
plt.axis(False)
for i in range(0, 5*8):
    plt.subplot(5, 8, i+1)
    plot_data_idx(rint(0, len(dataset["data"])-1))
```
